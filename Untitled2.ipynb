O‘zbekiston aholisi va migratsiya statistikasi (2017–2023)

Bu esa mamlakatdan ko‘chib ketayotganlarning qaytib kelayotganlardan ko‘pligini anglatadi. Masalan, 2017-yilda 1000 kishiga to‘g‘ri keladigan sof migratsiya darajasi -13 352 bo‘lgan bo‘lsa, 2023-yilga kelib bu ko‘rsatkich -6 004 gacha kamaygan.

12-bosqich — Korelatsiya tahlili
Bu bosqichda biz ustunlar orasidagi bog‘liqlikni (korelatsiya) hisoblaymiz. Masalan:

Remittances (% of GDP) bilan Net Migration orasida bog‘liqlik bormi?

Korelatsiya jadvali -1 dan +1 gacha qiymatlar beradi:

+1 → kuchli to‘g‘ri bog‘liqlik (birga ortadi).

-1 → kuchli teskari bog‘liqlik (biri oshsa, ikkinchisi kamayadi).

0 → bog‘liqlik yo‘q.



```python
import seaborn as sns

# Faqat sonli ustunlarni tanlaymiz
numeric_df = df.select_dtypes(include=["number"])

# Korelatsiya matritsasi
corr = numeric_df.corr()

# Jadval ko‘rish
print("Korelatsiya jadvali:")
print(corr)

# Issiqlik xarita (heatmap)
plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("O‘zbekiston statistik ko‘rsatkichlari orasidagi korelatsiya")
plt.show()
```

13-bosqich — Bashorat qilish (Forecast)
Bu bosqichda mavjud ma’lumotlar asosida kelajak uchun oddiy bashorat qilamiz. Masalan, 2024–2026 yillar uchun Net Migration va Remittances qiymatlarini oldindan taxmin qilish.

Oddiy usul sifatida Linear Regression (chiziqli regressiya) modelidan foydalanamiz. Bu usul yordamida tarixiy trend asosida kelajak qiymatlarini hisoblaymiz.

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Faqat yillar va remittances ustunini olamiz
X = df["Year"].values.reshape(-1, 1)
y = df["PersonalRemittancesofGDP"].values

# Modelni o‘qitamiz
model = LinearRegression()
model.fit(X, y)

# Kelajak yillar
future_years = np.array([2024, 2025, 2026]).reshape(-1, 1)
predictions = model.predict(future_years)

# Natijani chiqaramiz
for year, pred in zip(future_years.flatten(), predictions):
    print(f"{year} yil uchun bashorat qilingan Remittances (% of GDP): {pred:.2f}")

# Grafik ko‘rsatamiz
plt.figure(figsize=(10,6))
plt.scatter(X, y, color="blue", label="Haqiqiy ma’lumotlar")
plt.plot(X, model.predict(X), color="red", label="Regression chizig‘i")
plt.scatter(future_years, predictions, color="green", marker="x", s=100, label="Bashorat qilingan nuqtalar")

plt.title("Remittances (% of GDP) bashorati (2024–2026)")
plt.xlabel("Yil")
plt.ylabel("Remittances (% of GDP)")
plt.legend()
plt.grid(True)
plt.show()
```
